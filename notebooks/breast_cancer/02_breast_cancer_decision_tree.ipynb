# Eksperimen Decision Tree pada Breast Cancer Wisconsin Diagnostic Dataset

Notebook ini digunakan untuk menguji kestabilan performa algoritma Decision Tree terhadap variasi pembagian data training dan testing.

Eksperimen dilakukan menggunakan 30 nilai `random_state` yang berbeda, sedangkan konfigurasi model dan rasio train-test tetap sama.

```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)
```

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
MODEL_RANDOM_STATE = 42
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=42,
    stratify=y
)

model = DecisionTreeClassifier(
    random_state=MODEL_RANDOM_STATE
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
hasil_decision_tree = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    model = DecisionTreeClassifier(
        random_state=MODEL_RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_decision_tree.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

```python
df_hasil = pd.DataFrame(hasil_decision_tree)

df_hasil["accuracy_percent"] = (
    df_hasil["accuracy"] * 100
)

df_hasil.head()
```

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

```python
df_hasil.to_csv(
    "../../results/breast_cancer/breast_cancer_decision_tree.csv",
    index=False
)
```
