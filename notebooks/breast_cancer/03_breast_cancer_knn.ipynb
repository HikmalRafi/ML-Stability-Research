# Eksperimen K-Nearest Neighbors (KNN) pada Breast Cancer Wisconsin Diagnostic Dataset

Notebook ini digunakan untuk menguji kestabilan performa KNN terhadap variasi pembagian data training dan testing.

Eksperimen dilakukan menggunakan 30 nilai `random_state` yang berbeda dengan konfigurasi model dan rasio train-test yang tetap.

```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

data = load_breast_cancer()

X = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

y = pd.Series(
    data.target,
    name="target"
)
```

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
K = 5
```

## Scaling Data

KNN menghitung kedekatan antar-sampel berdasarkan jarak.

Karena fitur memiliki rentang nilai yang berbeda, StandardScaler digunakan agar fitur dengan skala numerik besar tidak mendominasi perhitungan jarak.

Scaler hanya dipelajari dari data training untuk mencegah data leakage.

```python
hasil_knn = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    model = KNeighborsClassifier(
        n_neighbors=K
    )

    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_knn.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

```python
df_hasil = pd.DataFrame(hasil_knn)

df_hasil["accuracy_percent"] = (
    df_hasil["accuracy"] * 100
)
```

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

```python
df_hasil.to_csv(
    "../../results/breast_cancer/breast_cancer_knn.csv",
    index=False
)
```
