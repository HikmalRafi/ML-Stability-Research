# Eksplorasi Breast Cancer Wisconsin Diagnostic Dataset

Notebook ini digunakan untuk memahami struktur Breast Cancer Wisconsin Diagnostic Dataset sebelum dilakukan eksperimen Machine Learning.

Tujuan:
- Memuat dataset.
- Melihat ukuran dan struktur data.
- Memahami fitur dan label.
- Memisahkan fitur (`X`) dan label (`y`).
- Memeriksa distribusi kelas.
- Memeriksa missing value.

```python
import pandas as pd

from sklearn.datasets import load_breast_cancer
```

## 1. Memuat Dataset

Breast Cancer Wisconsin Diagnostic Dataset tersedia secara langsung melalui library scikit-learn.

Dataset berisi sejumlah fitur numerik yang berasal dari karakteristik sel dan digunakan untuk membedakan dua kelas.

```python
data = load_breast_cancer()

type(data)
df = pd.DataFrame(
    data.data,
    columns=data.feature_names
)

df["target"] = data.target

df.head()
```

## 2. Ukuran Dataset

Ukuran dataset diperiksa untuk mengetahui jumlah sampel dan jumlah variabel yang tersedia.

```python
df.shape
```

## 3. Memisahkan Fitur dan Label

Fitur (`X`) berisi 30 variabel numerik yang digunakan oleh model untuk melakukan klasifikasi.

Label (`y`) berisi kelas sebenarnya dari setiap sampel.

```python
X = df.drop("target", axis=1)
y = df["target"]
```

## 4. Distribusi Kelas

Distribusi kelas diperiksa untuk mengetahui jumlah data pada masing-masing kelas.

```python
y.value_counts().sort_index()
```

```python
print("Malignant :", (y == 0).sum())
print("Benign    :", (y == 1).sum())
```

## 5. Pemeriksaan Missing Value

Missing value diperiksa karena data kosong dapat memengaruhi proses training dan evaluasi model.

```python
df.isnull().sum().sum()
```
