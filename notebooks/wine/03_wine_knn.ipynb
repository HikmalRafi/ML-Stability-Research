# Eksperimen K-Nearest Neighbors (KNN) pada Dataset Wine

Notebook ini digunakan untuk menguji performa algoritma K-Nearest Neighbors (KNN) pada Dataset Wine dengan variasi `random_state` pada pembagian data training dan testing.

Tujuan eksperimen:
- Menguji KNN pada Dataset Wine.
- Menggunakan rasio training dan testing yang tetap.
- Mengubah `random_state` dari 1 sampai 30.
- Menggunakan scaling agar setiap fitur berada pada skala yang sebanding.
- Mencatat perubahan akurasi pada setiap seed.
- Mengukur kestabilan performa KNN.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

## 1. Memuat Dataset

Dataset Wine dimuat kembali agar notebook ini dapat dijalankan secara mandiri.

Kolom `class` digunakan sebagai label, sedangkan 13 kolom lainnya digunakan sebagai fitur.

```python
columns = [
    "class",
    "alcohol",
    "malic_acid",
    "ash",
    "alcalinity_of_ash",
    "magnesium",
    "total_phenols",
    "flavanoids",
    "nonflavanoid_phenols",
    "proanthocyanins",
    "color_intensity",
    "hue",
    "od280_od315",
    "proline"
]

df = pd.read_csv(
    "../../data/wine/wine.data",
    header=None,
    names=columns
)

X = df.drop("class", axis=1)
y = df["class"]
```

## 2. Konfigurasi Eksperimen

Eksperimen menggunakan:

- 80% data untuk training.
- 20% data untuk testing.
- `stratify=y` untuk menjaga proporsi kelas.
- `random_state` pembagian data divariasikan dari 1 sampai 30.
- Nilai `K` pada KNN dibuat tetap agar perubahan hasil terutama berasal dari variasi pembagian data.

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
K = 5
```

## 3. Scaling Data

KNN menentukan kelas berdasarkan jarak antar-sampel.

Karena fitur pada Dataset Wine memiliki skala yang berbeda-beda, data perlu distandarisasi terlebih dahulu agar fitur dengan nilai numerik besar tidak mendominasi perhitungan jarak.

`StandardScaler` digunakan untuk melakukan standardisasi berdasarkan data training.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.transform(X_test)
```

## 4. Eksperimen Awal dengan Satu Seed

KNN diuji terlebih dahulu menggunakan satu pembagian data dengan `random_state=42`.

Nilai `K=5` digunakan sebagai konfigurasi awal.

```python
model = KNeighborsClassifier(n_neighbors=K)

model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

## 5. Eksperimen dengan 30 Random Seed

Eksperimen dilakukan sebanyak 30 kali dengan nilai `random_state` berbeda pada proses train-test split.

Konfigurasi KNN tetap sama pada setiap percobaan agar perubahan akurasi dapat diamati sebagai respons terhadap perubahan komposisi data training dan testing.

```python
hasil_knn = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    model = KNeighborsClassifier(
        n_neighbors=K
    )

    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_knn.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

## 6. Mengubah Hasil Eksperimen Menjadi Tabel

Hasil setiap seed diubah menjadi DataFrame agar lebih mudah dianalisis dan disimpan.

```python
df_hasil = pd.DataFrame(hasil_knn)

df_hasil["accuracy_percent"] = df_hasil["accuracy"] * 100

df_hasil.head()
```

## 7. Statistik Performa KNN

Statistik digunakan untuk melihat tingkat performa dan kestabilan KNN pada variasi train-test split.

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

## 8. Menyimpan Hasil Eksperimen

Hasil eksperimen disimpan dalam format CSV agar dapat digunakan kembali pada notebook perbandingan model.

```python
df_hasil.to_csv(
    "../../results/wine/wine_knn.csv",
    index=False
)
```

## Kesimpulan Sementara

Eksperimen KNN pada Dataset Wine menunjukkan bagaimana performa model berubah ketika komposisi data training dan testing diacak menggunakan random seed yang berbeda.
