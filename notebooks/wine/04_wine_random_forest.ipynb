# Eksperimen Random Forest pada Dataset Wine

Notebook ini digunakan untuk menguji performa algoritma Random Forest pada Dataset Wine dengan variasi `random_state` pada pembagian data training dan testing.

Tujuan eksperimen:
- Menguji Random Forest pada Dataset Wine.
- Menggunakan rasio training dan testing yang tetap.
- Mengubah `random_state` dari 1 sampai 30.
- Menjaga konfigurasi model tetap.
- Mencatat perubahan akurasi pada setiap seed.
- Mengukur kestabilan performa Random Forest.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

## 1. Memuat Dataset

Dataset Wine dimuat kembali agar notebook ini dapat dijalankan secara mandiri.

Kolom `class` digunakan sebagai label, sedangkan 13 kolom lainnya digunakan sebagai fitur.

```python
columns = [
    "class",
    "alcohol",
    "malic_acid",
    "ash",
    "alcalinity_of_ash",
    "magnesium",
    "total_phenols",
    "flavanoids",
    "nonflavanoid_phenols",
    "proanthocyanins",
    "color_intensity",
    "hue",
    "od280_od315",
    "proline"
]

df = pd.read_csv(
    "../../data/wine/wine.data",
    header=None,
    names=columns
)

X = df.drop("class", axis=1)
y = df["class"]
```

## 2. Konfigurasi Eksperimen

Eksperimen menggunakan:

- 80% data untuk training.
- 20% data untuk testing.
- `stratify=y` untuk menjaga proporsi kelas.
- `random_state` pembagian data divariasikan dari 1 sampai 30.
- `random_state` internal Random Forest dibuat tetap agar perubahan hasil terutama berasal dari variasi pembagian data.

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
MODEL_RANDOM_STATE = 42
N_ESTIMATORS = 100
```

## 3. Eksperimen Awal dengan Satu Seed

Random Forest diuji terlebih dahulu menggunakan satu pembagian data dengan `random_state=42`.

Model menggunakan 100 decision tree sebagai konfigurasi awal.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=42,
    stratify=y
)

model = RandomForestClassifier(
    n_estimators=N_ESTIMATORS,
    random_state=MODEL_RANDOM_STATE
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

## 4. Eksperimen dengan 30 Random Seed

Eksperimen dilakukan sebanyak 30 kali dengan nilai `random_state` berbeda pada proses pembagian data training dan testing.

Konfigurasi Random Forest tetap sama pada setiap percobaan agar perubahan akurasi dapat diamati sebagai respons terhadap perubahan komposisi data training dan testing.

```python
hasil_random_forest = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    model = RandomForestClassifier(
        n_estimators=N_ESTIMATORS,
        random_state=MODEL_RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_random_forest.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

## 5. Mengubah Hasil Eksperimen Menjadi Tabel

Hasil setiap seed diubah menjadi DataFrame agar lebih mudah dianalisis dan disimpan.

```python
df_hasil = pd.DataFrame(hasil_random_forest)

df_hasil["accuracy_percent"] = df_hasil["accuracy"] * 100

df_hasil.head()
```

## 6. Statistik Performa Random Forest

Statistik digunakan untuk melihat performa umum dan tingkat kestabilan Random Forest terhadap variasi pembagian data.

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

## 7. Menyimpan Hasil Eksperimen

Hasil eksperimen disimpan dalam format CSV agar dapat digunakan kembali pada tahap perbandingan model.

```python
df_hasil.to_csv(
    "../../results/wine/wine_random_forest.csv",
    index=False
)
```

## Kesimpulan Sementara

Eksperimen Random Forest pada Dataset Wine menunjukkan bagaimana performa model berubah ketika pembagian data training dan testing diacak menggunakan random seed yang berbeda.
