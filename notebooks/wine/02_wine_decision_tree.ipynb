# Eksperimen Decision Tree pada Dataset Wine

Notebook ini digunakan untuk menguji performa algoritma Decision Tree pada Dataset Wine dengan variasi `random_state` pada pembagian data training dan testing.

Tujuan eksperimen:
- Menguji Decision Tree pada Dataset Wine.
- Menggunakan rasio data training dan testing yang tetap.
- Mengubah `random_state` dari 1 sampai 30.
- Mencatat perubahan akurasi pada setiap seed.
- Mengukur kestabilan performa model.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

## 1. Memuat Dataset

Dataset Wine dimuat kembali karena setiap notebook dibuat agar dapat dijalankan secara mandiri.

Kolom `class` merupakan label, sedangkan kolom lainnya merupakan fitur yang digunakan oleh model.

```python
columns = [
    "class",
    "alcohol",
    "malic_acid",
    "ash",
    "alcalinity_of_ash",
    "magnesium",
    "total_phenols",
    "flavanoids",
    "nonflavanoid_phenols",
    "proanthocyanins",
    "color_intensity",
    "hue",
    "od280_od315",
    "proline"
]

df = pd.read_csv(
    "../../data/wine/wine.data",
    header=None,
    names=columns
)

X = df.drop("class", axis=1)
y = df["class"]
```

## 2. Konfigurasi Eksperimen

Eksperimen menggunakan:

- 80% data untuk training.
- 20% data untuk testing.
- `stratify=y` untuk menjaga proporsi kelas.
- `random_state` pembagian data divariasikan dari 1 sampai 30.
- `random_state` pada Decision Tree dibuat tetap agar perubahan hasil terutama berasal dari variasi pembagian data.

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
MODEL_RANDOM_STATE = 42
```

## 3. Eksperimen Awal dengan Satu Seed

Sebelum melakukan pengujian berulang, Decision Tree diuji terlebih dahulu menggunakan satu pembagian data dengan `random_state=42`.

Tahap ini digunakan untuk memastikan proses training dan prediksi berjalan dengan benar.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=42,
    stratify=y
)

model = DecisionTreeClassifier(
    random_state=MODEL_RANDOM_STATE
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy * 100:.2f}%")
```

## 4. Eksperimen dengan 30 Random Seed

Pada tahap ini, pembagian data training dan testing dilakukan sebanyak 30 kali.

Setiap percobaan menggunakan nilai `random_state` yang berbeda, sedangkan konfigurasi model dan rasio pembagian data tetap sama.

Tujuannya adalah untuk melihat apakah performa Decision Tree berubah ketika komposisi data training dan testing berubah.

```python
hasil_decision_tree = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    model = DecisionTreeClassifier(
        random_state=MODEL_RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_decision_tree.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

## 5. Mengubah Hasil Eksperimen Menjadi Tabel

Hasil setiap seed diubah menjadi DataFrame agar lebih mudah dianalisis dan disimpan.

```python
df_hasil = pd.DataFrame(hasil_decision_tree)

df_hasil.head()
```

```python
df_hasil["accuracy_percent"] = df_hasil["accuracy"] * 100

df_hasil.head()
```

## 6. Statistik Performa Decision Tree

Beberapa ukuran statistik digunakan untuk menggambarkan performa dan kestabilan model:

- Mean: rata-rata akurasi.
- Minimum: akurasi terendah.
- Maximum: akurasi tertinggi.
- Median: nilai tengah hasil eksperimen.
- Standard deviation: tingkat penyebaran akurasi dari rata-rata.
- Range: selisih antara hasil tertinggi dan terendah.

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

## 7. Menyimpan Hasil Eksperimen

Hasil eksperimen disimpan dalam format CSV agar dapat digunakan kembali pada tahap perbandingan model tanpa harus menjalankan seluruh eksperimen dari awal.

```python
df_hasil.to_csv(
    "../../results/wine_decision_tree.csv",
    index=False
)
```

## Kesimpulan Sementara

Eksperimen Decision Tree pada Dataset Wine menunjukkan bahwa perubahan pembagian data training dan testing dapat menghasilkan perubahan performa model.

Meskipun sebagian besar hasil berada pada tingkat akurasi yang relatif tinggi, terdapat beberapa seed yang menghasilkan akurasi lebih rendah maupun lebih tinggi.

Hal ini menunjukkan bahwa evaluasi model menggunakan hanya satu pembagian data belum tentu memberikan gambaran yang lengkap mengenai performa model.
