# Eksperimen Random Forest pada Raisin Dataset

Notebook ini digunakan untuk menguji kestabilan performa algoritma Random Forest terhadap variasi pembagian data training dan testing.

Eksperimen dilakukan menggunakan 30 nilai `random_state` yang berbeda, sedangkan konfigurasi model dan rasio train-test dibuat tetap.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

df = pd.read_excel(
    "../../data/raisin/Raisin_Dataset.xlsx"
)

X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
MODEL_RANDOM_STATE = 42
N_ESTIMATORS = 100
```

## Eksperimen dengan 30 Random Seed

Pembagian data training dan testing dilakukan sebanyak 30 kali dengan nilai `random_state` yang berbeda.

Konfigurasi Random Forest dibuat tetap agar perubahan hasil terutama mencerminkan perubahan komposisi data training dan testing.

```python
hasil_random_forest = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    model = RandomForestClassifier(
        n_estimators=N_ESTIMATORS,
        random_state=MODEL_RANDOM_STATE
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_random_forest.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

```python
df_hasil = pd.DataFrame(hasil_random_forest)

df_hasil["accuracy_percent"] = (
    df_hasil["accuracy"] * 100
)
```

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

```python
df_hasil.to_csv(
    "../../results/raisin/raisin_random_forest.csv",
    index=False
)
```

Eksperimen Random Forest pada Raisin Dataset menunjukkan bagaimana performa model berubah ketika komposisi data training dan testing divariasikan.
