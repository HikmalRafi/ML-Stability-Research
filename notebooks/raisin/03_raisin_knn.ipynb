# Eksperimen K-Nearest Neighbors (KNN) pada Raisin Dataset

Notebook ini digunakan untuk menguji kestabilan performa algoritma KNN terhadap variasi pembagian data training dan testing.

Eksperimen dilakukan menggunakan 30 nilai `random_state` yang berbeda dengan konfigurasi model dan rasio train-test yang tetap.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

df = pd.read_excel(
    "../../data/raisin/Raisin_Dataset.xlsx"
)

X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
TEST_SIZE = 0.2
JUMLAH_SEED = 30
K = 5
```

## Scaling Data

KNN melakukan klasifikasi berdasarkan jarak antar-sampel.

Karena fitur Raisin Dataset memiliki skala yang sangat berbeda, misalnya `Area` memiliki nilai puluhan ribu sedangkan `Eccentricity` berada di sekitar 0–1, standardisasi diperlukan agar fitur dengan angka besar tidak mendominasi perhitungan jarak.

StandardScaler hanya dipelajari dari data training untuk menghindari data leakage.

```python
hasil_knn = []

for seed in range(1, JUMLAH_SEED + 1):

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=seed,
        stratify=y
    )

    scaler = StandardScaler()

    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    model = KNeighborsClassifier(
        n_neighbors=K
    )

    model.fit(X_train_scaled, y_train)

    y_pred = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test, y_pred)

    hasil_knn.append({
        "seed": seed,
        "accuracy": accuracy
    })

    print(
        f"Seed {seed:2d} | "
        f"Accuracy: {accuracy * 100:.2f}%"
    )
```

```python
df_hasil = pd.DataFrame(hasil_knn)

df_hasil["accuracy_percent"] = (
    df_hasil["accuracy"] * 100
)
```

```python
mean_accuracy = df_hasil["accuracy_percent"].mean()
min_accuracy = df_hasil["accuracy_percent"].min()
max_accuracy = df_hasil["accuracy_percent"].max()
median_accuracy = df_hasil["accuracy_percent"].median()
std_accuracy = df_hasil["accuracy_percent"].std()
range_accuracy = max_accuracy - min_accuracy

print(f"Mean     : {mean_accuracy:.2f}%")
print(f"Minimum  : {min_accuracy:.2f}%")
print(f"Maximum  : {max_accuracy:.2f}%")
print(f"Median   : {median_accuracy:.2f}%")
print(f"Std Dev  : {std_accuracy:.2f}%")
print(f"Range    : {range_accuracy:.2f}%")
```

```python
df_hasil.to_csv(
    "../../results/raisin/raisin_knn.csv",
    index=False
)
```

Eksperimen KNN pada Raisin Dataset menunjukkan bagaimana performa model berubah ketika komposisi data training dan testing divariasikan.
